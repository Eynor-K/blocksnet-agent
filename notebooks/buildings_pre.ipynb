```python
import pandas as pd
import geopandas as gpd
import matplotlib.pyplot as plt
```

```python
df = gpd.read_file('data/raw/building.geojson')
```

```python
# Оставляем только те строки, где площадь (Area) находится в диапазоне от 0 до 100
df_filtered = df[(df['Area'] >= 0) & (df['Area'] <= 100)]

# Строим гистограмму распределения
plt.figure(figsize=(10, 6))
plt.hist(df_filtered['Area'].dropna(), bins=50, color='skyblue', edgecolor='black')

# Настраиваем заголовки и оси
plt.title('Распределение площади зданий (от 0 до 100)')
plt.xlabel('Площадь')
plt.ylabel('Количество зданий')
plt.grid(axis='y', alpha=0.75)

# Отображаем график (если запускаете локально)
plt.show()
```
