```python
import requests
import json

def download_wfs_layer():
    # Базовый URL геосервера
    base_url = "https://gisogd66.midural.ru/map/tiles/wfs"
    
    # Имя слоя, который мы хотим скачать
    layer_name = "agate_ekbgo:agate_ekbgo_geo_urban_10_functional_zone_950675df"
    
    # Параметры запроса (мы убрали cql_filter, чтобы скачать все объекты слоя)
    params = {
        "request": "GetFeature",
        "typeName": layer_name,
        "outputFormat": "application/json"
        # "maxFeatures": 5000 # Можно раскомментировать, если сервер будет отдавать ошибку из-за слишком большого объема данных
    }
    
    # Заголовки, чтобы притвориться обычным браузером
    headers = {
        "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36",
        "Accept": "application/json"
    }
    
    print(f"Начинаю загрузку слоя {layer_name}...")
    
    try:
        response = requests.get(base_url, params=params, headers=headers)
        
        # Проверяем, успешно ли прошел запрос
        if response.status_code == 200:
            data = response.json()
            
            # Проверяем, есть ли фичи в ответе
            if "features" in data:
                features_count = len(data["features"])
                print(f"Успешно скачано объектов: {features_count}")
                
                # Сохраняем результат в файл
                filename = "functional_zones.geojson"
                with open(filename, "w", encoding="utf-8") as f:
                    json.dump(data, f, ensure_ascii=False, indent=2)
                print(f"Данные сохранены в файл: {filename}")
            else:
                print("Сервер вернул ответ, но в нем нет объектов (features).")
                print("Ответ сервера:", data)
                
        else:
            print(f"Ошибка сервера! Статус код: {response.status_code}")
            print("Текст ошибки:", response.text)
            
    except Exception as e:
        print(f"Произошла ошибка при скачивании: {e}")

if __name__ == "__main__":
    download_wfs_layer()
```
