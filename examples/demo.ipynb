# BlocksNet Agent Demo

Минимальная демонстрация импорта Python-пакета `blocksnet_agent`.

Конфигурация берется из `.env`: `CHAT_URL` и `API_KEY`. Модель задается в ноутбуке через `MODEL`.

```python
MODEL = "gpt-4o-mini"
MAX_LOG_CHARS = 400
```

```python
import sys
from pathlib import Path

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == "examples" else Path.cwd()
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

from langchain_core.messages import AIMessage
from blocksnet_agent import BlocksNetAgent, AgentResult
```

```python
def message_tokens(message) -> int:
    usage = getattr(message, "usage_metadata", None) or {}
    if usage.get("output_tokens") is not None:
        return usage["output_tokens"]
    if usage.get("total_tokens") is not None:
        return usage["total_tokens"]
    return len(str(message.content).split())


def print_result(result: AgentResult, max_log_chars: int = MAX_LOG_CHARS) -> None:
    print(result["input"])
    print(result["output"])
    print(len(result["log"]))
    for i, message in enumerate(result["log"], start=1):
        if isinstance(message, AIMessage):
            print(f"\n--- message {i}: {type(message).__name__} | {message_tokens(message)} tokens ---")
        else:
            print(f"\n--- message {i}: {type(message).__name__} ---")
        content = str(message.content)
        if len(content) > max_log_chars:
            content = content[:max_log_chars] + " ...[обрезано]"
        print(content)
```

```python
agent = BlocksNetAgent(model=MODEL)
print("MODEL:", agent._settings.model)
print("CHAT_URL:", agent._settings.chat_url)
print("DATA_DIR:", agent._settings.data_dir)
print("OUTPUT_DIR:", agent._settings.output_dir)
```

## Пример 1

```python
result: AgentResult = agent.run(
    "Проанализируй транспортную доступность городских кварталов."
)
print_result(result)
```

## Пример 2

```python
result2 = agent.run("Оцени обеспеченность школами (15 мин).")
print_result(result2)
```

## Пример 3

```python
result3 = agent.run("Вычисли индекс Шеннона и центральность кварталов.")
print_result(result3)
```
